# Project Overview

The RAG-END Restaurant Management System is an AI-powered solution designed to streamline restaurant operations. It features an intelligent chatbot, QR code-based interactions, and a mobile-friendly interface to enhance customer experience and optimize business processes.

## Setup Environment

To set up the environment for the RAG-END project, follow these steps:

1. Ensure Python 3.9 or later is installed on your system.
2. Create a virtual environment:
   ```bash
   python -m venv env
   ```
3. Activate the virtual environment:
   - On Windows:
     ```bash
     .\env\Scripts\activate
     ```
   - On macOS/Linux:
     ```bash
     source env/bin/activate
     ```

## Install Dependencies

Once the virtual environment is activated, install the required dependencies:

```bash
pip install -r requirements.txt
```

## Run the Project

To start the RAG-END system, execute the following command:

```bash
python main.py
```

This will initialize the chatbot and other backend services required for the project.

## Test the Project

To verify that the project is working correctly, you can run the test suite:

```bash
python -m unittest discover tests
```

This will execute all unit tests and ensure that the core functionalities are working as expected.
